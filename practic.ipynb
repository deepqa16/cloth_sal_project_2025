```python
# get a fcatorial of input number
def factorial(n):
    if n == 0:
        return 1
    elif n<0:
        return "Please enter positive value:"
    else:
        return n*factorial(n-1)
n = int(input("Enter your no:"))
get_fact = factorial(n)
print(f"factorial of {n} is: {get_fact}")
```
